```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define two simple discrete-time signals
x = np.array([1, 2, 1, 1, 0, 3, 2, 1])
y = np.array([0, 1, 2, 1, 1, 0, 0, 0])

n_x = np.arange(len(x))
n_y = np.arange(len(y))

# 2. Compute Autocorrelation of x[n]
r_xx = np.correlate(x, x, mode='full')
lags_xx = np.arange(-len(x) + 1, len(x))

# 3. Compute Cross-correlation of x[n] and y[n]
r_xy = np.correlate(x, y, mode='full')
lags_xy = np.arange(-len(y) + 1, len(x))

# 4. Plot input signals individually
plt.figure(figsize=(10, 3))
plt.stem(n_x, x, linefmt='b-', markerfmt='bo', basefmt='r-')
plt.title("Signal x[n]")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 3))
plt.stem(n_y, y, linefmt='g-', markerfmt='go', basefmt='r-')
plt.title("Signal y[n]")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

# 5. Plot Autocorrelation and Cross-correlation
plt.figure(figsize=(10, 4))
plt.stem(lags_xx, r_xx, linefmt='C0-', markerfmt='C0o', label='Autocorrelation $r_{xx}$', basefmt='k-')
plt.stem(lags_xy, r_xy, linefmt='C1-', markerfmt='C1s', label='Cross-correlation $r_{xy}$', basefmt='k-')
plt.title("Comparison of Autocorrelation vs. Cross-correlation")
plt.xlabel("Lag (m)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()

# 6. Peak value analysis and comparison
peak_xx_val = np.max(r_xx)
peak_xx_lag = lags_xx[np.argmax(r_xx)]

peak_xy_val = np.max(r_xy)
peak_xy_lag = lags_xy[np.argmax(r_xy)]

print(f"Autocorrelation Peak: Value {peak_xx_val} at Lag {peak_xx_lag}")
print(f"Cross-correlation Peak: Value {peak_xy_val} at Lag {peak_xy_lag}")
```

```python
x = np.array([1, 2, 1, 1, 0, 3, 2, 1])
y = np.array([0, 1, 2, 1, 1, 0, 0, 0])

n_x = np.arange(len(x))
n_y = np.arange(len(y))

print("Signal x[n]:", x)
print("Signal y[n]:", y)
```

```python
# Individual visual for Signal x[n]
plt.figure(figsize=(10, 3))
plt.stem(n_x, x, linefmt='b-', markerfmt='bo', basefmt='r-')
plt.title("Individual Signal x[n]")
plt.xlabel("Sample index (n)")
plt.ylabel("Amplitude")
plt.show()
```

```python
# Individual visual for Signal y[n]
plt.figure(figsize=(10, 3))
plt.stem(n_y, y, linefmt='g-', markerfmt='go', basefmt='r-')
plt.title("Individual Signal y[n]")
plt.xlabel("Sample index (n)")
plt.ylabel("Amplitude")
plt.show()
```

```python
r_xx = np.correlate(x, x, mode='full')
lags_xx = np.arange(-len(x) + 1, len(x))

print("Autocorrelation r_xx:", r_xx)
print("Lags:", lags_xx)
```

```python
# Compute cross-correlation of x[n] and y[n]
r_xy = np.correlate(x, y, mode='full')
lags_xy = np.arange(-len(y) + 1, len(x))

print("Cross-correlation r_xy:", r_xy)
print("Lags:", lags_xy)
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True)

ax1.stem(n_x, x)
ax1.set_title("Signal x[n]")
ax1.set_ylabel("Amplitude")

ax2.stem(n_y, y)
ax2.set_title("Signal y[n]")
ax2.set_xlabel("Sample index (n)")
ax2.set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```

```python
# Plot autocorrelation and cross-correlation
fig, (ax1, ax2) = plt.subplots(2, 1)

ax1.stem(lags_xx, r_xx)
ax1.set_title("Autocorrelation $r_{xx}[m]$")
ax1.set_ylabel("Amplitude")

ax2.stem(lags_xy, r_xy)
ax2.set_title("Cross-correlation $r_{xy}[m]$")
ax2.set_xlabel("Lag (m)")
ax2.set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```

```python
# Combined overlay plot for side-by-side metric comparison
plt.figure(figsize=(10, 5))
plt.stem(lags_xx, r_xx, linefmt='C0-', markerfmt='C0o', label='Autocorrelation $r_{xx}$', basefmt='k-')
plt.stem(lags_xy, r_xy, linefmt='C1-', markerfmt='C1s', label='Cross-correlation $r_{xy}$', basefmt='k-')
plt.title("Visual Comparison: Autocorrelation vs. Cross-correlation")
plt.xlabel("Lag (m)")
plt.ylabel("Amplitude")
plt.legend()
plt.show()
```

```python
# Analyze the peak values and compare
peak_xx_idx = np.argmax(r_xx)
peak_xx_lag = lags_xx[peak_xx_idx]
peak_xx_val = r_xx[peak_xx_idx]

peak_xy_idx = np.argmax(r_xy)
peak_xy_lag = lags_xy[peak_xy_idx]
peak_xy_val = r_xy[peak_xy_idx]

print(f"Autocorrelation Peak: Value {peak_xx_val} at Lag {peak_xx_lag}")
print(f"Cross-correlation Peak: Value {peak_xy_val} at Lag {peak_xy_lag}")

# Quick explanation
print("\nComparison Summary:")
print("1. The autocorrelation peak always occurs at zero lag (lag = 0) because a signal aligned perfectly with itself maximizes similarity.")
print(f"2. The cross-correlation peak occurs at lag {peak_xy_lag}, indicating the relative shift at which the two signals are most closely aligned.")
```
