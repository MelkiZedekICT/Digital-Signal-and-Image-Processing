```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img_bgr = cv2.imread('/content/images.jpg')
if img_bgr is None:
    print("Error: Could not load image from /content/images.jpg")
else:
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

    plt.figure(figsize=(6, 6))
    plt.imshow(img_rgb)
    plt.title('Original Image')
    plt.axis('off')
    plt.show()
```

```python
_, threshold_img = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(6, 6))
plt.imshow(threshold_img, cmap='gray')
plt.title('Thresholded Image (Binary)')
plt.axis('off')
plt.show()
```

```python
negative_img = 255 - img_rgb

plt.figure(figsize=(6, 6))
plt.imshow(negative_img)
plt.title('Negative Image')
plt.axis('off')
plt.show()
```

```python
gamma = 1.8
# Normalize to [0, 1], apply gamma, and scale back to [0, 255]
gamma_img = np.array(255 * (img_rgb / 255.0) ** gamma, dtype='uint8')

plt.figure(figsize=(6, 6))
plt.imshow(gamma_img)
plt.title(f'Gamma Corrected Image (Gamma = {gamma})')
plt.axis('off')
plt.show()
```
